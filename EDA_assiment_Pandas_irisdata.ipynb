```python
import numpy as np
import pandas as pd
import seaborn as sns

df = sns.load_dataset('iris')

# Copy dataset
df_transformed = df.copy()

# Log Transformation
df_transformed['sepal_length_log'] = np.log(df['sepal_length'])

# Square Root Transformation
df_transformed['sepal_length_sqrt'] = np.sqrt(df['sepal_length'])

# Square Transformation
df_transformed['sepal_length_square'] = df['sepal_length'] ** 2

# Cube Transformation
df_transformed['sepal_length_cube'] = df['sepal_length'] ** 3

print(df_transformed.head())
```

```python
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# Load iris dataset
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)

print("First 5 rows of Iris Dataset")
print(df.head())


# Mean and Standard Deviation before standardization
print("\nBefore Standardization")
print("Mean:\n", df.mean())
print("Standard Deviation:\n", df.std())

# Min-Max Normalization
minmax = MinMaxScaler()
normalized_data = minmax.fit_transform(df)

df_normalized = pd.DataFrame(normalized_data, columns=iris.feature_names)

print("\nAfter Min-Max Normalization")
print(df_normalized.head())

# Standardization (Z-score)
scaler = StandardScaler()
standardized_data = scaler.fit_transform(df)

df_standardized = pd.DataFrame(standardized_data, columns=iris.feature_names)

print("\nAfter Standardization")
print(df_standardized.head())

# Mean and Standard Deviation after standardization
print("\nAfter Standardization Statistics")
print("Mean:\n", df_standardized.mean())
print("Standard Deviation:\n", df_standardized.std())
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

# Load Iris dataset
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = iris.target

# 1. Boxplot for 3 numerical columns
plt.figure(figsize=(8,5))
sns.boxplot(data=df[['sepal length (cm)','sepal width (cm)','petal length (cm)']])
plt.title("Boxplot for Numerical Features")
plt.show()

# 2. Identify Outliers using IQR method
for col in ['sepal length (cm)','sepal width (cm)','petal length (cm)']:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(f"\nOutliers in {col}:")
    print(outliers[col])

# 3. Pairplot
sns.pairplot(df, hue='species')
plt.show()
```

```python
Section G: Interpretation

1. Which column has highest skewness?
In the Iris dataset, petal width (cm) usually shows the highest skewness compared to other columns. This means the values are not evenly distributed and are slightly shifted toward one side of the distribution. Skewness indicates how much the data deviates from a perfectly symmetrical distribution.

2. Which transformation worked best?
Standardization (Z-score) worked best because it converts the data so that the mean becomes 0 and standard deviation becomes 1. This transformation is useful in many machine learning algorithms because it keeps the distribution shape while scaling the values.

3. Did normalization change the distribution?
No, normalization does not change the distribution shape of the data. It only rescales the values to a range between 0 and 1. The pattern and relationships in the data remain the same.

4. What insights did you get from the pairplot?

Petal length and petal width show a strong positive relationship.

Sepal length has a moderate relationship with petal length.

Sepal width has a weak relationship with other variables.

The three Iris species form distinct clusters, especially based on petal measurements.
```
